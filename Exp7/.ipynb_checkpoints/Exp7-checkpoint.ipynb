Prepare the academic performance sheet for college which takes the performance of 30 students based on their study hours, attendance, and examination marks. Using Python, calculate the mean, median, variance, and standard deviation of marks, and determine the correlation between study hours and marks. Plot appropriate graphs, including a scatter plot and normal distribution curve, and interpret the results to understand student performance and data distribution.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("StudentPerformanceFactors.csv")
```

```python
df.head()
```

```python
df.info()
```

```python
sns.heatmap(df.isnull(), cmap="viridis", yticklabels=False)
```

```python
sns.boxplot(data=df, y='Exam_Score')
```

```python
print("Mean of the Student Marks",df['Exam_Score'].mean())
print("Median of the Student Marks",df['Exam_Score'].median())
print("Variance of the Student Marks",df['Exam_Score'].var())
print("Std Dev of the Student Marks",df['Exam_Score'].var()**0.5)
```

```python
df['Hours_Studied'].corr(df['Exam_Score'])
plt.figure(figsize=(8,5))

plt.scatter(df['Hours_Studied'], df['Exam_Score'])

plt.xlabel('Hours Studied')
plt.ylabel('Exam Score')
plt.show()
```

```python
sns.scatterplot(data=df, x="Attendance", y="Exam_Score")
```

```python
plt.figure(figsize=(8,5))

sns.histplot(df['Exam_Score'], kde=True)

plt.xlabel('Exam Score')
plt.ylabel('Number of Students')
plt.title('Distribution of Examination Marks')

plt.show()
```

```python
sns.boxplot(data=df, x="School_Type",y="Exam_Score")
```
