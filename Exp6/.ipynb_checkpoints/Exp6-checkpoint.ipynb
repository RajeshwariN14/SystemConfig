A financial analyst wants to study the historical performance of Tesla (TSLA) using daily stock-market data. Write a Python program to find the highest and lowest closing prices and their corresponding dates, Calculate the daily percentage change in closing price. Display the 10 days having the highest daily gain and Plot the closing-price trend using Matplotlib.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("tesla_data.csv")
```

```python
df.head()
```

```python
high=df['Close'].max()
ind=df['Close'].idxmax()

dateHigh=df.loc[ind,'Date']
print("Highest Closing Price:",high)
print("Highest Closing Price Recorded on:",dateHigh)
```

```python
low=df['Close'].min()
ind2=df['Close'].idxmin()

dateLow=df.loc[ind2,'Date']
print("Lowest Closing Price:",low)
print("Lowest Closing Price Recorded on:",dateLow)
```

```python
df['change%']=df['Close'].pct_change()*100

df.head()
```

```python
top_10=df.sort_values(by=['change%'], ascending=False)

print("10 days having the highest daily gain")
arr=pd.DataFrame(top_10[["Date","change%"]])
```

```python
arr.head(10)
```

```python
plt.figure(figsize=(12,4))

plt.plot(df['Date'], df['Close'])
plt.xlabel('Date')
plt.ylabel('Closing Price')
```
