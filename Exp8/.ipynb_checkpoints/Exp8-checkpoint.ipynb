A real-estate company wants to analyze house prices, carpet area, and number of bedrooms for 50 houses. Using Python, calculate the mean, median, variance, and standard deviation of house prices, study the correlation between carpet area and price, and visualize the data using scatter plots and a histogram with a normal distribution curve. Interpret the results and determine whether house prices are approximately normally distributed and whether larger houses tend to have higher prices.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import norm
```

```python
df=pd.read_csv("Housing.csv")
```

```python
df.head()
```

```python
print("Mean",df['price'].mean())
print("Median",df['price'].median())
print("Std Dev",(df['price'].var()) ** 0.5)
```

```python
df['area'].corr(df['price'])
```

```python
sns.scatterplot(data=df, x='area', y='price', hue='furnishingstatus')
```

```python
sns.heatmap(data=df.corr(numeric_only=True), annot=True, cmap='coolwarm')
```

```python
sns.histplot(df['price'], bins=20, kde=False, stat='density')
```
